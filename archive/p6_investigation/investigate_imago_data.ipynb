```python
# parse project data p6 phenotype
import pandas_plink
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from os.path import join
```

## CTP_201 HAPLOTYPE CALLER

```python
f = '/path/to/local_data/p6/investigate_sample_data/CTP_201/germline_call_bam/06_GDB/chr9/chr9'
snp_info,sample_info,genotypes  = pandas_plink.read_plink(f)
genotype_mat = genotypes.compute()
print(genotype_mat.shape)
print(snp_info.head)
print(sample_info)

# fix the switching of alleles
gen_sub = genotype_mat * -1 + 2
print(gen_sub.shape)
# gen_sub = np.nan_to_num(gen_sub)
print(gen_sub[:,0:15])


n_genotypes = np.product(gen_sub.shape)
genotyping_rate = (n_genotypes - np.sum(np.isnan(gen_sub))) / n_genotypes
print(n_genotypes)
print(genotyping_rate)

gen_sub_df = pd.DataFrame(gen_sub)
# sample_names = [x.split(".")[0][10:-5] for x in sample_info.iid]
# print(sample_info.iid[0:10])
# print(sample_names[0:10])
gen_sub_df.columns = sample_info.iid
gen_sub_df.index = snp_info.pos
gen_sub_df.to_csv('/path/to/local_data/p6/investigate_sample_data/CTP_201/p6_parsed_haplotypecaller.tsv', sep='\t')
```
